```python

from openai import OpenAI
from IPython.display import Markdown, display, update_display
import openai
import ollama


```

Build the prompt

```python
openai_model = "gpt-4o-mini"
ollama_model = "llama3.2"
system_prompt = "You are a football commentator."
system_prompt += " You provide funny, light hearted, encouraging, impartial updates to games based on requests."
system_prompt += " Always name the teams in your response."
#system_prompt += "Always look for the team logos and use them in the response"
system_prompt += "You will always get team names, competition name, group name (if any) pitch."
system_prompt += "You will also get other game events like, game started, half time and game ended, which team just scored"
system_prompt += "Keep your responses to max 150 characters"
system_prompt += "Your response is always in markdown."
system_prompt += "The information you get is a list of all your responses so far. You need to figure out what the next response should be."

#system_prompt += "Answer in Finnish"
#system_prompt += "Answer in Swedish"

#system_prompt = "You are a football commentator providing light, funny, encouraging, impartial updates with team names, competition, pitch, and events (e.g. goals, halftime, start, end). Max 200 characters."

#print(system_prompt)

user_prompt = ""
user_prompt += "Competition P2019. Group 1. Pitch 4. BK-46 vs EIF has just kicked off."
#user_prompt += "Competition P2019. Group 1. Pitch 4. BK-46 vs EIF. BK-46 scores. New score BK-46 1 : EIF : 0"
#user_prompt += "Competition P2019. Group 1. Pitch 4. BK-46 vs EIF. EIF scores. New score BK-46 1 : EIF : 1"
#user_prompt += "Competition P2019. Group 1. Pitch 4. BK-46 vs EIF. BK-46 scores. New score BK-46 2 : EIF : 1"
#user_prompt += "Competition P2019. Group 1. Pitch 4. BK-46 vs EIF. EIF scores. New score BK-46 2 : EIF : 2"
#user_prompt += "Competition P2019. Group 1. Pitch 4. BK-46 vs EIF. EIF scores. New score BK-46 2 : EIF : 3."
#user_prompt += "Competition P2019. Group 1. Pitch 4. Game over. Final score: BK-46 2 : EIF : 3."

messages = [
    {"role": "system", "content": system_prompt},
    {"role": "user", "content": user_prompt}
]
```

Build the API call to OpenAI

```python

response = openai.chat.completions.create(
    model=openai_model,
    messages=messages
)

display(Markdown(response.choices[0].message.content))
```

Build the Call to Ollama

```python
response = ollama.chat(
    model=ollama_model,
    messages=messages
)

display(Markdown(response['message']['content']))
```
