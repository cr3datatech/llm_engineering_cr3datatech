Summary of all the major learnings from Week 1

```python

from openai import OpenAI
from IPython.display import Markdown, display, update_display
import openai
import ollama


```

Build the prompt

```python
openai_model = "gpt-4o-mini"
ollama_model = "llama3.2"
system_prompt = "You are my PHP guru. I need you to guide me through the PHP implementation of everything I ask you"
user_prompt = "Show me how to connect to OpenAI API. Return the result in markdown"

messages = [
    {"role": "system", "content": system_prompt},
    {"role": "user", "content": user_prompt}
]
```

Build the API call to OpenAI

```python

response = openai.chat.completions.create(
    model=openai_model,
    messages=messages
)

display(Markdown(response.choices[0].message.content))
```

Build the Call to Ollama

```python
response = ollama.chat(
    model=ollama_model,
    messages=messages
)

display(Markdown(response['message']['content']))
```
