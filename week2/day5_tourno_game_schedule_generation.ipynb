```python

from openai import OpenAI
from IPython.display import Markdown, display, update_display
import openai
import ollama
```

Somewhere to store prompts

```python
def test_user_prompts():
    return [
        "Competition P2019. Group 1. Pitch 4. BK-46 vs EIF has just kicked off.",
        "BK-46 scores.",
        #"BK-46 scores.",
        "BK-46 scores.",
        "EIF scores.",
        #"EIF scores.",
        #"EIF scores.",
        "Game Over."
    ]

USER_PROMPT = ""  # Global variable
USER_PROMPT_COUNT = 0  # Global variable

def add_to_user_prompt(text=None):
    global USER_PROMPT
    global USER_PROMPT_COUNT
    if text:
        USER_PROMPT += text
    else:
        if(len(test_user_prompts()) > USER_PROMPT_COUNT):
            USER_PROMPT += test_user_prompts()[USER_PROMPT_COUNT]
            USER_PROMPT_COUNT += 1

system_prompt = "You are a football tournament shedule specialist."
system_prompt += " Your audience is parents and kids"
system_prompt += " You provide funny, light hearted, playful, encouraging, impartial updates to games based on requests."
system_prompt += " Keep your responses to max 150 characters including emojis."
system_prompt += " Your response is a single line of text and always in markdown."
system_prompt += " Always name the competition, pitch and teams."
system_prompt += " Include the game score if not kickoff."
system_prompt += " Always include the game score if game over."

```

Build the prompt

```python
openai_model = "gpt-4o-mini"
ollama_model = "llama3.2"
global system_prompt
global USER_PROMPT
global USER_PROMPT_COUNT

def create_messages():
    messages = [
        {"role": "system", "content": system_prompt},
        {"role": "user", "content": USER_PROMPT}
    ]
    return messages
```

Build the API call to OpenAI

```python

add_to_user_prompt()
response = openai.chat.completions.create(
    model=openai_model,
    messages=create_messages()
)

display(Markdown(response.choices[0].message.content))
add_to_user_prompt(response.choices[0].message.content)

print(USER_PROMPT)
```

Build the Call to Ollama

```python
add_to_user_prompt()
response = ollama.chat(
    model=ollama_model,
    messages=create_messages()
)

display(Markdown(response['message']['content']))
add_to_user_prompt(response['message']['content'])

print(USER_PROMPT)
```
